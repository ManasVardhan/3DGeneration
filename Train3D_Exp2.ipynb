```python
"""
Colab Cell: Visualize Y values (Ground Truth 3D Mesh) for any shoe
Copy this entire cell into Google Colab
"""

# ============================================================================
# VISUALIZATION CELL - Copy this into Colab
# ============================================================================

import numpy as np
import trimesh
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from pathlib import Path

# Configuration
SHOE_ID = "2"  # ← Change this to visualize different shoes
OBJ_DIR = "/content/3DGeneration/data/OBJs"  # Adjust path if needed

# ============================================================================
# Find and Load the OBJ file
# ============================================================================

print("="*70)
print(f"VISUALIZING Y VALUES FOR SHOE ID: {SHOE_ID}")
print("="*70)

# Find OBJ file for this shoe
obj_paths = list(Path(OBJ_DIR).rglob("*.obj"))
shoe_obj_path = None

for obj_path in obj_paths:
    shoe_folder_id = obj_path.parent.name
    if shoe_folder_id == SHOE_ID:
        shoe_obj_path = obj_path
        break

if shoe_obj_path is None:
    print(f"❌ ERROR: No OBJ file found for Shoe ID '{SHOE_ID}'")
    print(f"\nAvailable shoe IDs:")
    available_ids = sorted(set([p.parent.name for p in obj_paths]))
    for sid in available_ids[:20]:
        print(f"  - {sid}")
    if len(available_ids) > 20:
        print(f"  ... and {len(available_ids) - 20} more")
else:
    print(f"✓ Found OBJ: {shoe_obj_path}")

    # Load mesh
    print("\nLoading mesh...")
    mesh = trimesh.load(shoe_obj_path, process=False, force='mesh')

    vertices = mesh.vertices
    faces = mesh.faces

    print(f"✓ Loaded mesh:")
    print(f"  Vertices: {len(vertices):,}")
    print(f"  Faces: {len(faces):,}")

    # Extract colors if available
    has_colors = False
    vertex_colors = None

    if hasattr(mesh.visual, 'vertex_colors'):
        vertex_colors = mesh.visual.vertex_colors[:, :3] / 255.0
        has_colors = True
        print(f"  Colors: Yes (vertex colors)")
    elif hasattr(mesh.visual, 'material'):
        try:
            material = mesh.visual.material
            if hasattr(material, 'diffuse'):
                color = material.diffuse[:3] / 255.0
                vertex_colors = np.tile(color, (len(vertices), 1))
                has_colors = True
                print(f"  Colors: Yes (material diffuse: {color})")
        except:
            pass

    if not has_colors:
        print(f"  Colors: No (will use white)")
        vertex_colors = np.ones((len(vertices), 3))

    # Normalize mesh to [-1, 1] (same as training data)
    print("\nNormalizing to [-1, 1] cube (same as training)...")
    centroid = vertices.mean(axis=0)
    vertices_normalized = vertices - centroid
    max_dist = np.abs(vertices_normalized).max()
    if max_dist > 0:
        vertices_normalized = vertices_normalized / max_dist

    print(f"  Original bbox: [{vertices.min(axis=0)}, {vertices.max(axis=0)}]")
    print(f"  Normalized bbox: [{vertices_normalized.min(axis=0)}, {vertices_normalized.max(axis=0)}]")

    # ========================================================================
    # VISUALIZATION 1: 3D Mesh with Multiple Views
    # ========================================================================

    print("\nGenerating visualizations...")

    fig = plt.figure(figsize=(20, 10))

    # View angles (similar to the 6 camera views)
    views = [
        ("Front View", 0, 0),
        ("Back View", 0, 180),
        ("Left View", 0, -90),
        ("Right View", 0, 90),
        ("Top View", 90, 0),
        ("Bottom View", -90, 0)
    ]

    for idx, (title, elev, azim) in enumerate(views, 1):
        ax = fig.add_subplot(2, 3, idx, projection='3d')

        # Plot mesh
        ax.plot_trisurf(
            vertices_normalized[:, 0],
            vertices_normalized[:, 1],
            vertices_normalized[:, 2],
            triangles=faces,
            cmap='viridis',
            alpha=0.9,
            edgecolor='none',
            shade=True
        )

        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.view_init(elev=elev, azim=azim)
        ax.set_xlim([-1, 1])
        ax.set_ylim([-1, 1])
        ax.set_zlim([-1, 1])
        ax.set_box_aspect([1, 1, 1])

    plt.suptitle(f'Y Values (Ground Truth Mesh) - Shoe ID: {SHOE_ID}',
                 fontsize=16, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.show()

    # ========================================================================
    # VISUALIZATION 2: Point Cloud with Colors
    # ========================================================================

    fig = plt.figure(figsize=(15, 5))

    # Sample points for point cloud visualization
    num_sample_points = min(8192, len(vertices_normalized))
    sample_indices = np.random.choice(len(vertices_normalized), num_sample_points, replace=False)
    sample_points = vertices_normalized[sample_indices]
    sample_colors = vertex_colors[sample_indices] if has_colors else None

    views_simple = [
        ("3D View 1", 30, 45),
        ("3D View 2", 30, 135),
        ("3D View 3", 60, 225)
    ]

    for idx, (title, elev, azim) in enumerate(views_simple, 1):
        ax = fig.add_subplot(1, 3, idx, projection='3d')

        if has_colors and sample_colors is not None:
            ax.scatter(
                sample_points[:, 0],
                sample_points[:, 1],
                sample_points[:, 2],
                c=sample_colors,
                s=1,
                alpha=0.6
            )
        else:
            ax.scatter(
                sample_points[:, 0],
                sample_points[:, 1],
                sample_points[:, 2],
                c=sample_points[:, 2],  # Color by Z coordinate
                cmap='viridis',
                s=1,
                alpha=0.6
            )

        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.view_init(elev=elev, azim=azim)
        ax.set_xlim([-1, 1])
        ax.set_ylim([-1, 1])
        ax.set_zlim([-1, 1])
        ax.set_box_aspect([1, 1, 1])

    plt.suptitle(f'Point Cloud Representation ({num_sample_points} points) - Shoe ID: {SHOE_ID}',
                 fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

    # ========================================================================
    # STATISTICS
    # ========================================================================

    print("\n" + "="*70)
    print("MESH STATISTICS (Y VALUES)")
    print("="*70)
    print(f"Shoe ID: {SHOE_ID}")
    print(f"OBJ Path: {shoe_obj_path}")
    print(f"\nGeometry:")
    print(f"  Vertices: {len(vertices):,}")
    print(f"  Faces: {len(faces):,}")
    print(f"  Edges: {len(mesh.edges):,}")
    print(f"\nBounding Box (Original):")
    print(f"  Min: {vertices.min(axis=0)}")
    print(f"  Max: {vertices.max(axis=0)}")
    print(f"  Size: {vertices.max(axis=0) - vertices.min(axis=0)}")
    print(f"\nBounding Box (Normalized to [-1, 1]):")
    print(f"  Min: {vertices_normalized.min(axis=0)}")
    print(f"  Max: {vertices_normalized.max(axis=0)}")
    print(f"\nColors:")
    print(f"  Has vertex colors: {has_colors}")
    if has_colors:
        print(f"  Color range: [{vertex_colors.min():.3f}, {vertex_colors.max():.3f}]")
        print(f"  Mean color: {vertex_colors.mean(axis=0)}")
    print(f"\nMesh Properties:")
    print(f"  Is watertight: {mesh.is_watertight}")
    print(f"  Is convex: {mesh.is_convex}")
    print(f"  Volume: {mesh.volume:.6f}")
    print(f"  Surface area: {mesh.area:.6f}")
    print("="*70)

print("\n✓ Visualization complete!")
print("\nTo visualize a different shoe, change SHOE_ID at the top of this cell.")
```

```python
"""
Colab Cell: Visualize X values (Input Images) - What the model receives
Copy this entire cell into Google Colab
"""

# ============================================================================
# VISUALIZATION CELL - Copy this into Colab
# ============================================================================

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
import torch
import torchvision.transforms as transforms

# Configuration
SHOE_ID = "2"  # ← Change this to visualize different shoes
IMAGES_DIR = "/content/3DGeneration/data/input_images"  # Adjust path if needed

# ============================================================================
# Load the 6 multi-view images
# ============================================================================

print("="*70)
print(f"VISUALIZING X VALUES (INPUT IMAGES) FOR SHOE ID: {SHOE_ID}")
print("="*70)

views = ['front', 'back', 'left', 'right', 'top', 'bottom']
images_dir = Path(IMAGES_DIR)

# Find and load images
images_raw = {}
images_processed = {}

print(f"\nLoading images from: {images_dir}")

for view in views:
    # Try different extensions
    img_path = None
    for ext in ['.png', '.jpeg', '.jpg']:
        test_path = images_dir / f"{SHOE_ID}_{view}{ext}"
        if test_path.exists():
            img_path = test_path
            break

    if img_path is None:
        print(f"❌ Image not found: {SHOE_ID}_{view}.[png|jpeg|jpg]")
        print(f"\nSearching for available images with ID '{SHOE_ID}'...")
        matching_files = list(images_dir.glob(f"{SHOE_ID}_*"))
        if matching_files:
            print(f"Found {len(matching_files)} files:")
            for f in matching_files:
                print(f"  - {f.name}")
        else:
            print(f"No files found matching '{SHOE_ID}_*'")
            # Show available shoe IDs
            all_images = list(images_dir.glob("*_front.*"))
            if all_images:
                available_ids = sorted(set([f.stem.rsplit('_', 1)[0] for f in all_images]))
                print(f"\nAvailable shoe IDs (found {len(available_ids)}):")
                for sid in available_ids[:20]:
                    print(f"  - {sid}")
                if len(available_ids) > 20:
                    print(f"  ... and {len(available_ids) - 20} more")
        break
    else:
        # Load image
        img = Image.open(img_path).convert('RGB')
        images_raw[view] = img
        print(f"✓ Loaded {view}: {img.size} - {img_path.name}")

if len(images_raw) != 6:
    print(f"\n❌ ERROR: Expected 6 images, found {len(images_raw)}")
else:
    print(f"\n✓ All 6 views loaded successfully!")

    # ========================================================================
    # Process images (same as model input)
    # ========================================================================

    print("\nProcessing images (same as model preprocessing)...")

    # DINOv2 expects 512x512 images
    transform = transforms.Compose([
        transforms.Resize((512, 512)),
        transforms.ToTensor(),  # Converts to [0, 1] and (C, H, W)
    ])

    for view, img in images_raw.items():
        img_tensor = transform(img)
        images_processed[view] = img_tensor
        print(f"  {view}: {img_tensor.shape} (C, H, W), range [{img_tensor.min():.3f}, {img_tensor.max():.3f}]")

    # ========================================================================
    # VISUALIZATION 1: Raw Input Images (6 Views)
    # ========================================================================

    print("\nGenerating visualizations...")

    fig = plt.figure(figsize=(20, 10))

    for idx, view in enumerate(views, 1):
        ax = fig.add_subplot(2, 3, idx)
        ax.imshow(images_raw[view])
        ax.set_title(f'{view.upper()} VIEW', fontsize=14, fontweight='bold')
        ax.axis('off')

        # Add image info
        img_array = np.array(images_raw[view])
        ax.text(0.02, 0.98,
                f'Size: {images_raw[view].size}\nShape: {img_array.shape}',
                transform=ax.transAxes,
                fontsize=10,
                verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.suptitle(f'X Values (Model Input) - Raw Images - Shoe ID: {SHOE_ID}',
                 fontsize=16, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.show()

    # ========================================================================
    # VISUALIZATION 2: Processed Images (After Preprocessing)
    # ========================================================================

    fig = plt.figure(figsize=(20, 10))

    for idx, view in enumerate(views, 1):
        ax = fig.add_subplot(2, 3, idx)

        # Convert tensor back to displayable format
        img_display = images_processed[view].permute(1, 2, 0).cpu().numpy()

        ax.imshow(img_display)
        ax.set_title(f'{view.upper()} VIEW (Processed)', fontsize=14, fontweight='bold')
        ax.axis('off')

        # Add tensor info
        tensor = images_processed[view]
        ax.text(0.02, 0.98,
                f'Tensor: {tuple(tensor.shape)}\nRange: [{tensor.min():.3f}, {tensor.max():.3f}]',
                transform=ax.transAxes,
                fontsize=10,
                verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.suptitle(f'X Values (Model Input) - Processed (512x512, Normalized) - Shoe ID: {SHOE_ID}',
                 fontsize=16, fontweight='bold', y=0.98)
    plt.tight_layout()
    plt.show()

    # ========================================================================
    # VISUALIZATION 3: Individual Channel Analysis (RGB)
    # ========================================================================

    # Pick one view to analyze in detail
    analyze_view = 'front'

    fig = plt.figure(figsize=(20, 5))

    img_tensor = images_processed[analyze_view]

    # Original RGB
    ax = fig.add_subplot(1, 4, 1)
    ax.imshow(img_tensor.permute(1, 2, 0).cpu().numpy())
    ax.set_title('RGB Combined', fontsize=12, fontweight='bold')
    ax.axis('off')

    # R channel
    ax = fig.add_subplot(1, 4, 2)
    ax.imshow(img_tensor[0].cpu().numpy(), cmap='Reds')
    ax.set_title(f'R Channel\nRange: [{img_tensor[0].min():.3f}, {img_tensor[0].max():.3f}]',
                 fontsize=11)
    ax.axis('off')

    # G channel
    ax = fig.add_subplot(1, 4, 3)
    ax.imshow(img_tensor[1].cpu().numpy(), cmap='Greens')
    ax.set_title(f'G Channel\nRange: [{img_tensor[1].min():.3f}, {img_tensor[1].max():.3f}]',
                 fontsize=11)
    ax.axis('off')

    # B channel
    ax = fig.add_subplot(1, 4, 4)
    ax.imshow(img_tensor[2].cpu().numpy(), cmap='Blues')
    ax.set_title(f'B Channel\nRange: [{img_tensor[2].min():.3f}, {img_tensor[2].max():.3f}]',
                 fontsize=11)
    ax.axis('off')

    plt.suptitle(f'Channel Analysis - {analyze_view.upper()} View - Shoe ID: {SHOE_ID}',
                 fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

    # ========================================================================
    # STATISTICS
    # ========================================================================

    print("\n" + "="*70)
    print("INPUT DATA STATISTICS (X VALUES)")
    print("="*70)
    print(f"Shoe ID: {SHOE_ID}")
    print(f"Images directory: {images_dir}")
    print(f"\nImage Details:")

    for view in views:
        img = images_raw[view]
        tensor = images_processed[view]
        img_array = np.array(img)

        print(f"\n  {view.upper()} VIEW:")
        print(f"    Raw image size: {img.size}")
        print(f"    Raw array shape: {img_array.shape}")
        print(f"    Processed tensor shape: {tuple(tensor.shape)} (C, H, W)")
        print(f"    Pixel value range (raw): [0, 255]")
        print(f"    Tensor value range (processed): [{tensor.min():.4f}, {tensor.max():.4f}]")
        print(f"    Mean RGB: {tensor.mean(dim=[1, 2]).cpu().numpy()}")
        print(f"    Std RGB: {tensor.std(dim=[1, 2]).cpu().numpy()}")

    print(f"\nPreprocessing Pipeline:")
    print(f"  1. Resize to 512x512 (DINOv2 input size)")
    print(f"  2. Convert to PyTorch tensor")
    print(f"  3. Normalize to [0, 1] range")
    print(f"  4. Channel order: RGB (C, H, W)")

    print(f"\nModel Input Format:")
    print(f"  Batch size: 1 (during training)")
    print(f"  Tensor shape per view: (1, 3, 512, 512)")
    print(f"  Total views: 6")
    print(f"  Device: {torch.device('cuda' if torch.cuda.is_available() else 'cpu')}")

    # Stack all views to show batched format
    stacked = torch.stack([images_processed[v] for v in views])
    print(f"\nStacked tensor shape (all views): {tuple(stacked.shape)}")
    print(f"  (num_views, channels, height, width)")

    print("="*70)

    print("\n✓ Visualization complete!")
    print("\nTo visualize a different shoe, change SHOE_ID at the top of this cell.")
```
